```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/Loan.csv")
```

```python
print("Rows:",df.shape[0])
```

```python
print("Columns:", df.shape[1])
print(df.columns.tolist())
df.info()
```

```python
print("Duplicate rows:", df.duplicated().sum())
print(df["LoanApproved"].value_counts())
print(df["RiskScore"].describe())
```

```python
cat_cols = df.select_dtypes(include="object").columns
for col in cat_cols:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
import matplotlib.pyplot as plt

df["LoanApproved"].value_counts().plot(kind="bar", color=["steelblue","orange"])
plt.title("Loan Approval Distribution")
plt.xlabel("Loan Approved (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.savefig('images/loan_approval_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
plt.hist(df["RiskScore"], bins=30, color="mediumseagreen", edgecolor="black")
plt.title("Risk Score Distribution")
plt.xlabel("Risk Score")
plt.ylabel("Frequency")
plt.savefig('images/risk_score_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(14,10))
corr = df.select_dtypes(include=["int64","float64"]).corr()
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.savefig('images/correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
print("Top correlations with RiskScore:")
print(corr["RiskScore"].sort_values(ascending=False))

print("\nTop correlations with LoanApproved:")
print(corr["LoanApproved"].sort_values(ascending=False))
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="LoanApproved", y="RiskScore", data=df, palette=["orange","steelblue"])
plt.title("Risk Score by Loan Approval Status")
plt.xlabel("Loan Approved (0 = No, 1 = Yes)")
plt.ylabel("Risk Score")
plt.savefig('images/risk_score_by_approval.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="LoanApproved", y="TotalDebtToIncomeRatio", data=df, hue="LoanApproved", palette=["orange","steelblue"], legend=False)
plt.title("Debt-to-Income Ratio by Loan Approval Status")
plt.xlabel("Loan Approved (0 = No, 1 = Yes)")
plt.ylabel("Total Debt-to-Income Ratio")
plt.savefig('images/debt_to_income_by_approval.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
approval_by_employment = df.groupby("EmploymentStatus")["LoanApproved"].mean().sort_values(ascending=False) * 100

plt.figure(figsize=(8,5))
approval_by_employment.plot(kind="bar", color="teal")
plt.title("Loan Approval Rate by Employment Status")
plt.xlabel("Employment Status")
plt.ylabel("Approval Rate (%)")
plt.xticks(rotation=45)
plt.savefig('images/approval_by_employment.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
print(df["EmploymentStatus"].value_counts())
print(df.groupby("EmploymentStatus")["LoanApproved"].mean())
```

```python
def risk_category(score):
    if score < 45:
        return "Low Risk"
    elif score < 60:
        return "Medium Risk"
    else:
        return "High Risk"

df["RiskCategory"] = df["RiskScore"].apply(risk_category)
print(df["RiskCategory"].value_counts())
```

```python
plt.figure(figsize=(8,5))
order = ["Low Risk", "Medium Risk", "High Risk"]
sns.countplot(x="RiskCategory", data=df, order=order, palette=["green","orange","red"], hue="RiskCategory", legend=False)
plt.title("Applicant Distribution by Risk Category")
plt.savefig('images/applicant_by_risk_category.png', dpi=150, bbox_inches='tight')
plt.show()

print(df.groupby("RiskCategory")["LoanApproved"].mean().reindex(order) * 100)
```

```python
df.to_csv("../data/processed/loan_data_with_risk.csv", index=False)
print("Saved successfully!")
```
